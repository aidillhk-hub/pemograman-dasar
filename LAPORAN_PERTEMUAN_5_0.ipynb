```python
# 1. Parameter awal
jumlah_buku = 7
harga_per_buku = 18500
persentase_diskon = 0.10
uang_budi = 150000
jumlah_orang_patungan = 4

# Perhitungan dasar
total_belanja = jumlah_buku * harga_per_buku
besar_diskon = total_belanja * persentase_diskon
jumlah_harus_dibayar = total_belanja - besar_diskon
uang_kembalian = uang_budi - jumlah_harus_dibayar

# Perhitungan patungan
iuran_per_orang = int(jumlah_harus_dibayar // jumlah_orang_patungan)
sisa_iuran = int(jumlah_harus_dibayar % jumlah_orang_patungan)

# Status kecukupan uang
uang_cukup = uang_budi >= jumlah_harus_dibayar

# Menampilkan hasil
print("=== HASIL PERHITUNGAN SOAL 1 ===")
print(f"Total Belanja: Rp {total_belanja:,.0f}")
print(f"Besar Diskon (10%): Rp {besar_diskon:,.0f}")
print(f"Jumlah yang Harus Dibayar: Rp {jumlah_harus_dibayar:,.0f}")
print(f"Uang Kembalian Budi: Rp {uang_kembalian:,.0f}")
print(f"Iuran per orang (4 orang): Rp {iuran_per_orang:,.0f}")
print(f"Sisa iuran pembulatan: Rp {sisa_iuran:,.0f}")
print(f"Apakah uang Budi cukup?: {uang_cukup}")
```

```python
# 1. Parameter Input
jumlah_panel = 12
daya_per_panel_wp = 450
jam_penyinaran = 5
efisiensi = 0.8
kebutuhan_harian_kwh = 25
tarif_per_kwh = 1444.70

# Perhitungan
daya_total_kw = (jumlah_panel * daya_per_panel_wp) / 1000
energi_harian_kwh = daya_total_kw * jam_penyinaran * efisiensi

energi_bulanan_kwh = energi_harian_kwh * 30
nilai_rupiah_bulanan = energi_bulanan_kwh * tarif_per_kwh

kekurangan_harian_kwh = max(0, kebutuhan_harian_kwh - energi_harian_kwh)

apakah_mencukupi = energi_harian_kwh >= kebutuhan_harian_kwh

# Menampilkan Hasil
print("=== HASIL PERHITUNGAN PLTS ===")
print(f"Energi per hari: {energi_harian_kwh:.1f} kWh")
print(f"Energi per bulan (30 hari): {energi_bulanan_kwh:.1f} kWh")
print(f"Nilai rupiah listrik per bulan: Rp {nilai_rupiah_bulanan:,.2f}")
print(f"Kekurangan energi per hari: {kekurangan_harian_kwh:.1f} kWh")
print(f"Apakah PLTS sudah mencukupi?: {apakah_mencukupi}")
```

```python
# Parameter input
daya_generator_kw = 200
jam_operasi = 8
konsumsi_solar_per_kwh = 0.25
harga_solar_per_liter = 13500
kapasitas_tangki_liter = 1000
arus_ampere = 20
hambatan_ohm = 0.5

# 1. Energi yang dibangkitkan per hari (kWh)
energi_harian_kwh = daya_generator_kw * jam_operasi

# 2. Kebutuhan solar per hari dan biayanya
kebutuhan_solar_hari_liter = energi_harian_kwh * konsumsi_solar_per_kwh
biaya_solar_harian = kebutuhan_solar_hari_liter * harga_solar_per_liter

# 3. Lama hari tangki penuh bertahan dan sisa solar
hari_tangki_cukup = int(kapasitas_tangki_liter // kebutuhan_solar_hari_liter)
sisa_solar_liter = kapasitas_tangki_liter % kebutuhan_solar_hari_liter

# 4. Rugi daya pada kabel (P = I^2 * R)
rugi_daya_watt = (arus_ampere ** 2) * hambatan_ohm

# 5. Apakah rugi kabel tergolong tinggi? (> 150 W dan arus >= 20 A)
apakah_rugi_tinggi = (rugi_daya_watt > 150) and (arus_ampere >= 20)

# Menampilkan hasil
print("=== HASIL PERHITUNGAN GENERATOR & KABEL ===")
print(f"1. Energi dibangkitkan per hari  : {energi_harian_kwh:,.0f} kWh")
print(f"2. Kebutuhan solar per hari      : {kebutuhan_solar_hari_liter:,.0f} liter")
print(f"   Biaya solar harian            : Rp {biaya_solar_harian:,.0f}")
print(f"3. Tangki cukup untuk            : {hari_tangki_cukup} hari penuh")
print(f"   Sisa solar di tangki          : {sisa_solar_liter:,.0f} liter")
print(f"4. Rugi daya pada kabel          : {rugi_daya_watt:,.0f} Watt")
print(f"5. Apakah rugi kabel tinggi?      : {apakah_rugi_tinggi}")
```
