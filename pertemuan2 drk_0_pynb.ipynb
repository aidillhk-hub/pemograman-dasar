```python
# Variabel untuk panel surya
tegangan_panel_surya = 18
arus_panel_surya = 6
jenis_sel_panel_surya = "Monocrystalline"

# Mencetak nilai dari setiap variabel
print(f"Tegangan panel surya: {tegangan_panel_surya} volt")
print(f"Arus panel surya: {arus_panel_surya} ampere")
print(f"Jenis sel panel surya: {jenis_sel_panel_surya}")
```

```python
# Variabel dengan tipe int (Resistor)
nilai_resistor = 1000

# Variabel dengan tipe float (Kapasitor)
nilai_kapasitor = 0.000001

# Variabel dengan tipe bool (LED menyala)
led_menyala = True

# Variabel dengan tipe str (Mikrokontroler)
jenis_mikrokontroler = "Arduino Uno"

# Mencetak nilai dan tipe dari setiap variabel
print(f"Nilai resistor: {nilai_resistor}, Tipe: {type(nilai_resistor)}")
print(f"Nilai kapasitor: {nilai_kapasitor}, Tipe: {type(nilai_kapasitor)}")
print(f"LED menyala: {led_menyala}, Tipe: {type(led_menyala)}")
print(f"Jenis mikrokontroler: {jenis_mikrokontroler}, Tipe: {type(jenis_mikrokontroler)}")
```

```python
# Variabel untuk informasi PLTS Nunukan
nama_unit = "PLTS Nunukan"
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = True

# Mencetak nilai dan tipe dari setiap variabel
print(f"Nama unit: {nama_unit}, Tipe: {type(nama_unit)}")
print(f"Jumlah panel: {jumlah_panel}, Tipe: {type(jumlah_panel)}")
print(f"Tegangan keluaran: {tegangan_keluaran}, Tipe: {type(tegangan_keluaran)}")
print(f"Sedang beroperasi: {sedang_beroperasi}, Tipe: {type(sedang_beroperasi)}")
```

```python
# Daftar pembacaan tegangan harian
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print(f"Pembacaan pertama: {tegangan_harian[0]} V")
print(f"Pembacaan terakhir: {tegangan_harian[-1]} V")

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)
print(f"Daftar setelah ditambahkan: {tegangan_harian}")

# (c) Cetak jumlah total data dengan len()
print(f"Jumlah total data: {len(tegangan_harian)}")
```

```python
# Buat dictionary spesifikasi motor listrik
motor_listrik = {
    "nama": "Motor Induksi AC",
    "daya_kW": 15.0,
    "tegangan_V": 380,
    "rpm": 1450
}

# Cetak salah satu nilai lewat key-nya (misalnya, daya_kW)
print(f"Daya motor: {motor_listrik['daya_kW']} kW")

# Ubah nilai rpm
motor_listrik['rpm'] = 1500

# Cetak ulang seluruh dictionary
print("\nSpesifikasi motor listrik setelah perubahan RPM:")
print(motor_listrik)
```

```python
# Inisialisasi daftar unit pembangkit
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"}
]

# (a) Cetak nama unit kedua
print(f"Nama unit kedua: {pembangkit[1]['nama']}")

# (b) Ubah status salah satu unit (misal: Unit-3 menjadi 'siaga')
pembangkit[2]['status'] = "siaga"
print(f"Status Unit-3 setelah diubah: {pembangkit[2]['status']}")

# (c) Tambahkan satu unit baru ke list dengan append()
unit_baru = {"nama": "Unit-4", "kapasitas_MW": 120, "status": "rencana"}
pembangkit.append(unit_baru)
print(f"\nUnit baru ditambahkan: {unit_baru['nama']}")

# (d) Cetak seluruh data unit pembangkit setelah perubahan
print("\nSeluruh data unit pembangkit:")
for unit in pembangkit:
    print(unit)
```
